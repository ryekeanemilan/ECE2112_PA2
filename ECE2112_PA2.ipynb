**Reproducible Normalization Problem**

```python
import numpy as np
```

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
mean = np.mean(X)
std = np.std(X)

X_normalized = (X-mean)/std


print("Original Array X:\n", X)
print("\nNormalized Array X:\n", X_normalized)
print("\nMean of normalized array:", X_normalized.mean())
print("\nStandard deviation of normalized array:", X_normalized.std())
np.save('X_normalized.npy', X_normalized)
```

**Cubes Divisible by 4 Problem**

```python
C = np.power(np.arange (1,101), 3).reshape(10, 10)
C

div_by_4 = C[np.mod(C, 4) == 0]
div_by_4

print("Shape of Array C:", C.shape)
print("\nElements divisible by 4:\n", div_by_4)
print("\nNumber of selected elements:", div_by_4.size)
np.save('div_by_4.npy', div_by_4)
```

**Above-Mean Squares Problem**

```python
S = np.power(np.arange (1,37), 2).reshape(6, 6)
S

S_mean = np.mean(S)
above_mean = S[S>S_mean]
print("Array S:\n", S)
print("\nElements above the mean:\n", above_mean)
print("\nNumber of selected elements:", above_mean.size)
print("\nMean of S:", S_mean)
np.save('above_mean.npy', above_mean)
```
